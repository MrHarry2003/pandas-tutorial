```python
import pandas as pd
```

```python
orders_data = pd.read_csv('data/orders.csv')
```

```python
orders_data.head()
```

```python
country = orders_data['Country'] # Gives a series of all the countries in the data
country
```

```python
country_with_customer = orders_data[['CustomerName', 'Country']] # Gives a DataFrame with only the CustomerName and Country columns
country_with_customer
```

```python
row_one = orders_data.iloc[0] # Gives the first row of the data
row_one
```

```python
# Applying a filter -- Just write the condition in the square brackets to filter the data
electronic_products = orders_data[(orders_data['Category'] == 'Electronics') & (orders_data['Country'] == 'USA')] # Use single | for OR condition
electronic_products
```

```python
customer_with_a = orders_data[orders_data['CustomerName'].str.startswith('A')] # Filters the data for all the customers whose name starts with A
customer_with_a
```

```python
orders_data[orders_data['Country'].isin(['USA', 'Canada'])] # Filters the data for all the customers whose country is either USA or Canada
```

```python
print(orders_data.loc[orders_data['CustomerName'] == 'Alex Johnson']['Product'])
orders_data.loc[orders_data['CustomerName'] == 'Alex Johnson', 'Product'] = 'Tim' # Updates the product name for the customer Alex Johnson to Tim
orders_data[orders_data['CustomerName'] == 'Alex Johnson']
```

```python
orders_data['Country'].value_counts() # Gives the count of all the unique values in the Country column
```

```python
orders_data.groupby(['Country','Category'])['Price'].sum() 
```
